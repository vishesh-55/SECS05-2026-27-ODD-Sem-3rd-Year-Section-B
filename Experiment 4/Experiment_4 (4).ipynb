```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/workspaces/-Mastery_in_Data-Analytics_and_Visualizations_and_Career-Advancement/WA_Fn-UseC_-HR-Employee-Attrition-selected-columns.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Department")

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=20)

plt.show()
```

```python
line_data = df.groupby("Age")["DailyRate"].mean()

plt.figure(figsize=(10, 5))

plt.plot(
    line_data.index,
    line_data.values,
    marker="o"
)

plt.title("Average Daily Rate by Age")
plt.xlabel("Age")
plt.ylabel("Average Daily Rate")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Age"],
    bins=15,
    edgecolor="black"
)

plt.title("Distribution of Employee Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Department",
    y="Age"
)

plt.title("Age Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="DailyRate"
)

plt.title("Age vs Daily Rate")
plt.xlabel("Age")
plt.ylabel("Daily Rate")

plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

ax = sns.countplot(
    data=df,
    x="EducationField"
)

plt.title("Employees by Education Field", fontsize=16)
plt.xlabel("Education Field", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)

plt.xticks(rotation=30)

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```

```python
print("========== DATA VISUALIZATION SUMMARY ==========")

print("\n1. Bar Chart:")
print("Used to compare employees across different departments.")

print("\n2. Line Chart:")
print("Used to observe the trend of average Daily Rate with Age.")

print("\n3. Histogram:")
print("Used to understand the distribution of employee Age.")

print("\n4. Box Plot:")
print("Used to compare Age distributions across departments and identify outliers.")

print("\n5. Scatter Plot:")
print("Used to analyze the relationship between Age and Daily Rate.")

print("\n6. Correlation Heatmap:")
print("Used to visualize correlations between numerical variables.")

print("\nOverall Conclusion:")
print("Data visualization makes patterns, trends, distributions,")
print("relationships and possible outliers easier to understand.")
```
