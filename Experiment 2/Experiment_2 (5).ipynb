```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

file_path = "/workspaces/-Mastery_in_Data-Analytics_and_Visualizations_and_Career-Advancement/WA_Fn-UseC_-HR-Employee-Attrition-selected-columns.csv"

df = pd.read_csv(file_path)

print(df.head())
```

```python
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nSummary Statistics:")
display(df.describe())

print("\nAttrition Value Counts:")
print(df["Attrition"].value_counts())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Age"], kde=True)

plt.title("Distribution of Employee Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Department")

plt.title("Employee Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.xticks(rotation=20)
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

correlation_matrix = df[numeric_cols].corr()

display(correlation_matrix)

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(y=df["Age"])

plt.title("Box Plot of Employee Age")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="DailyRate",
    hue="Attrition"
)

plt.title("Age vs Daily Rate")
plt.xlabel("Age")
plt.ylabel("Daily Rate")

plt.show()
```

```python
# Univariate Analysis
print("UNIVARIATE ANALYSIS")
print("\nAge Statistics:")
print(df["Age"].describe())

print("\nAttrition Distribution:")
print(df["Attrition"].value_counts())

# Bivariate Analysis
print("\nBIVARIATE ANALYSIS")
print("\nAverage Age by Attrition:")
print(df.groupby("Attrition")["Age"].mean())

print("\nAverage Daily Rate by Attrition:")
print(df.groupby("Attrition")["DailyRate"].mean())

# Multivariate Analysis
print("\nMULTIVARIATE ANALYSIS")
display(
    df.groupby(["Department", "Attrition"])["Age"].mean().reset_index()
)
```

```python
print("========== EDA FINDINGS ==========")

print("\n1. Dataset contains", df.shape[0], "records and", df.shape[1], "features.")

print("\n2. Attrition distribution:")
print(df["Attrition"].value_counts())

print("\n3. Average Age by Attrition:")
print(df.groupby("Attrition")["Age"].mean())

print("\n4. Average Daily Rate by Attrition:")
print(df.groupby("Attrition")["DailyRate"].mean())

print("\n5. EDA helps identify:")
print("- Data distributions")
print("- Relationships between variables")
print("- Outliers and anomalies")
print("- Frequently occurring categories")
print("- Important patterns for business decisions")

print("\nCONCLUSION:")
print("Exploratory Data Analysis provides a better understanding of the dataset")
print("and helps identify useful patterns before statistical analysis or machine learning.")
```
