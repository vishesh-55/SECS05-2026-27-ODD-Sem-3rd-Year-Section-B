Theory Questions    1. What is classification, and how does it differ from regression?    Classification   is a supervised learning technique   used to predict   categories or classes   ,   such as Yes/No or Churn/Not Churn.    Regression   is used to predict a   continuous numerical   value   , such as salary, price, or   temperature.    Example:    ●     Classification → Customer will churn: Yes/No   ●     Regression → Predict customer spending amount.    2. Explain the working principle of the Decision Tree algorithm.    A Decision Tree makes predictions by dividing the dataset into smaller groups using a series   of   decision rules   .

The algorithm:   1.     Selects the best feature for splitting.   2.     Divides the data into branches.   3.     Repeats the process for each branch.   4.     Stops when a suitable stopping condition is reached.   5.     Uses the final leaf node to make the prediction.   It uses criteria such as   Gini Index or Entropy   to select the best split.    3. Differentiate between Gini Index and Entropy as splitting criteria.    Gini Index     Entropy    Measures impurity in a dataset.     Measures uncertainty or disorder.    Generally faster to calculate.     Uses logarithmic calculations.    Lower Gini means a better split.     Lower Entropy means a better split.    Commonly used with Decision Trees.     Used for information-gain based splitting.    Both are used to select suitable splits in a Decision Tree.    4. What is a Confusion Matrix? Explain its components.    A   Confusion Matrix   is a table used to evaluate the   performance of a classification model.   Its four components are:   ●     True Positive (TP):   Correctly predicted positive cases.   ●     True Negative (TN):   Correctly predicted negative cases.   ●     False Positive (FP):   Negative case incorrectly predicted   as positive.   ●     False Negative (FN):   Positive case incorrectly predicted   as negative.   It helps understand the types of errors made by the model.    5. Define Accuracy, Precision, Recall, and F1-Score. Why are these   metrics important?    ●     Accuracy:   Percentage of total predictions that are   correct.    Accuracy = (TP + TN) / (TP + TN + FP + FN)

--- End of Page 8 ---

●     Precision:   Percentage of predicted positive cases   that are actually positive.    Precision = TP / (TP + FP)    ●     Recall:   Percentage of actual positive cases correctly   identified.    Recall = TP / (TP + FN)    ●     F1-Score:   Harmonic mean of Precision and Recall.    F1 = 2 × (Precision × Recall) / (Precision + Recall)    These metrics are important because they provide a detailed view of classification   performance rather than depending only on accuracy.    6. What is overfitting in a Decision Tree? How can it be reduced?    Overfitting   occurs when a Decision Tree learns the   training data too closely, including noise   and unnecessary details. As a result, it performs well on training data but poorly on new   data.   It can be reduced by:   ●     Limiting the   maximum tree depth   .   ●     Setting a minimum number of samples for splitting.   ●     Pruning unnecessary branches.   ●     Using suitable train-test splitting and validation.    7.
