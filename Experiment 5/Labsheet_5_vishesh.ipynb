Theory Questions    1. What is Business Intelligence (BI)? How does it support organizational   decision making?    Business Intelligence (BI) is the process of collecting, analyzing, and presenting data to help   organizations make better decisions. BI converts raw data into meaningful information using   reports, dashboards, and visualizations. It helps managers identify trends, monitor   performance, reduce risks, and make data-driven decisions.    2. What is an interactive dashboard? How is it different from a static   report?    An interactive dashboard is a visual display of data where users can interact with charts,   filters, slicers, and other elements. A static report mainly presents fixed information that does   not change based on user interaction. Interactive dashboards allow users to explore data   and get specific insights quickly.    3. ●     Conversion Rate   ●     Customer Acquisition Cost    5. Explain the purpose of filters, slicers, and drill-down functionality in   dashboards.    Filters   are used to display only selected data based   on specific conditions.   Slicers   provide a   visual way to filter dashboard data, such as selecting a year, region, or product.   Drill-down    allows users to move from summarized information to more detailed information, such as   Year → Quarter → Month.    6. What factors should be considered while designing an effective   business dashboard?    An effective dashboard should consider:   ●     Clear and simple design   ●     Relevant KPIs and data   ●     Easy-to-understand charts   ●     Proper use of colors and labels   ●     Interactive filters and slicers   ●     Fast performance   ●     Target audience and business objectives   The dashboard should present important information without unnecessary complexity.    7. How can dashboards help managers monitor organizational   performance in real time?    Dashboards can connect to regularly updated data sources and display the latest   information. Managers can monitor sales, revenue, expenses, inventory, and employee   performance from a single screen. Alerts and KPIs can also highlight problems quickly,   helping managers take timely corrective actions.    8. Why is data visualization important in Business Intelligence tools?    Data visualization represents complex data using charts, graphs, maps, and dashboards. It   makes large amounts of information easier to understand. Visualization helps users identify   trends, patterns, relationships, and unusual values quickly, which improves decision-making.    9.
